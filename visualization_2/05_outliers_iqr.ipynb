Problema 5: Atrasos de voos - IQR e limites

```python
import pandas as pd

url = "https://raw.githubusercontent.com/Rogerio-mack/Visualizacao-de-Dados-em-Python/main/data/flights_delays_2015.csv"
flights = pd.read_csv(url)

# a) Escolhendo a primeira do top 5
comp_escolhida = flights['name'].value_counts().index[0]
dados_comp = pd.to_numeric(flights.loc[flights['name'] == comp_escolhida, 'arr_delay'], errors='coerce').dropna()

# b, c e d) Cálculos
Q1 = dados_comp.quantile(0.25)
Mediana = dados_comp.median()
Q3 = dados_comp.quantile(0.75)
IQR = Q3 - Q1
LI = Q1 - 1.5 * IQR
LS = Q3 + 1.5 * IQR

# e, f e g) Identificação
outliers = dados_comp[(dados_comp < LI) | (dados_comp > LS)]
perc = (len(outliers) / len(dados_comp)) * 100

print(f"Limites - Inferior: {LI} | Superior: {LS}")
print(f"Potenciais outliers: {len(outliers)} ({perc:.2f}%)")

# i) Média vs Mediana
media_antes = dados_comp.mean()
media_depois = dados_comp[(dados_comp >= LI) & (dados_comp <= LS)].mean()
print(f"Média antes: {media_antes:.2f} | Média depois: {media_depois:.2f}")
print(f"Mediana (inalterada): {Mediana}")
```

h) Exclusão automática? Não. Atrasos de voos severos não são erros de inserção, são fenómenos reais da operação logística. Excluí-los destrói dados essenciais para gerir riscos.

i) Efeito da retirada: A média reduziu significativamente (porque valores altos extremos exerciam enorme peso na soma). A mediana, por ser robusta, não sofreu praticamente alteração.
