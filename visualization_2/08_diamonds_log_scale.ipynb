Problema 8: Diamantes - escala logarítmica

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

url_diamonds = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/diamonds.csv"
diamonds = pd.read_csv(url_diamonds)

diamonds['log_price'] = np.log10(diamonds['price'])

# Gráficos com Escala Linear vs Logarítmica
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

sns.histplot(data=diamonds, x='price', ax=axes[0,0], bins=50)
axes[0,0].set_title("Histograma (Escala Linear)")

sns.histplot(data=diamonds, x='log_price', ax=axes[0,1], bins=50)
axes[0,1].set_title("Histograma (Escala Log10)")

sns.boxplot(data=diamonds, x='cut', y='price', ax=axes[1,0])
axes[1,0].set_title("Box-plot por Cut (Linear)")

sns.violinplot(data=diamonds, x='cut', y='log_price', ax=axes[1,1])
axes[1,1].set_title("Violin Plot por Cut (Log10)")

plt.tight_layout()
plt.show()
```

b) Assimetria: O histograma em escala linear mostra uma severa assimetria direita.

d) Efeito da transformação: A escala logarítmica "encurta" o tamanho das distâncias na cauda longa, normalizando o formato da distribuição para um sino mais visível (ainda que neste dataset com alguma bimodalidade logarítmica).

h) Comparação por cut: A qualidade "Ideal" surpreendentemente detém densidade acentuada nos preços mais baixos. Níveis "Premium" estendem-se por avaliações mais elevadas.

i) Utilidade da Escala Logarítmica: Quando ordens de grandeza são dispares e a variação não é linear e sim percentual/multiplicativa (como no luxo e finanças).

j) Informar o leitor: Deve ser obrigatoriamente informada! Um leigo assume que as distâncias visuais crescem na mesma proporção em todos os quadrantes, podendo tirar conclusões errôneas no acompanhamento de crescimento se a escala for logarítmica.
