Desafio Opcional: Como uma distribuição pode enganar?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/Rogerio-mack/Visualizacao-de-Dados-em-Python/main/data/flights_delays_2015.csv"
arr_delay = pd.to_numeric(pd.read_csv(url)['arr_delay'], errors='coerce').dropna()

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Pouquíssimas classes (mascara padrões e picos secundários)
sns.histplot(arr_delay, bins=5, ax=axes[0,0])
axes[0,0].set_title("Hist: bins=5 (Ilusão centralizada)")

# 2. Muitas classes (Gera ruído massivo e aparente caos de variações)
sns.histplot(arr_delay, bins=500, ax=axes[0,1])
axes[0,1].set_title("Hist: bins=500 (Ruído estocástico)")

# 3. KDE com bandwidth minúsculo (Cria múltiplas mini-distribuições falsas)
sns.kdeplot(arr_delay, bw_adjust=0.01, ax=axes[1,0])
axes[1,0].set_title("KDE: bw_adjust=0.01 (Serrilhado falso)")

# 4. KDE gigante (Esmaga padrões criando um sino perfeito falso)
sns.kdeplot(arr_delay, bw_adjust=10, ax=axes[1,1])
axes[1,1].set_title("KDE: bw_adjust=10 (Suavização excessiva)")

plt.tight_layout()
plt.show()
```

Conclusão: Escolhas supostamente "técnicas" afetam e subvertem severamente a interpretação humana dos dados. bins minúsculos criam uma ilusão monolítica central ocultando a existência da cauda. Um bandwidth gigante num KDE suaviza as curvas num modelo gaussiano imaculado que elimina a natureza comportamental humana subjacente ao atraso do aeroporto. Em contrapartida, bins massivos e bw baixíssimos geram ruído branco e picos "dentes-de-serra", fazendo crer erroneamente na existência de dezenas de clusters distintos. Dominar as parametrizações define se o relatório traduz ciência real ou apenas poluição visual arbitrária.
