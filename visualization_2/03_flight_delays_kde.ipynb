Problema 3: Atrasos de voos - KDE e efeito do bandwidth

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/Rogerio-mack/Visualizacao-de-Dados-em-Python/main/data/flights_delays_2015.csv"
arr_delay = pd.to_numeric(pd.read_csv(url)['arr_delay'], errors='coerce').dropna()

# a) e b) Histograma Normalizado com KDE
sns.histplot(x=arr_delay, stat="density", bins=100, kde=True)
plt.show()

# c) KDE com diferentes bandwidths
fig, ax = plt.subplots(figsize=(8,5))
for bw in [0.3, 1, 2]:
    sns.kdeplot(arr_delay, bw_adjust=bw, label=f'bw_adjust={bw}')
ax.legend()
plt.show()

# f) Selecionando atrasos reais
atrasos = arr_delay[arr_delay > 0]
sns.kdeplot(atrasos, bw_adjust=1)
plt.show()
```

d) Efeito do bandwidth: O parâmetro ajusta a suavização. Valores baixos seguem cada ponto gerando ruído; valores altos fundem picos ocultando a variabilidade real.

e) Avaliação: Irregular: bw_adjust=0.3. Equilibrada: bw_adjust=1. Excessivamente suavizada: bw_adjust=2.

g) Valores menores que zero: Acontece porque a KDE aplica uma função matemática contínua e suave (como uma Gaussiana) sobre cada dado. Pontos muito próximos de zero projetam uma "cauda" matemática em valores negativos, sendo apenas um artefacto visual da curva.

h) Interpretação: A KDE deve ser vista como uma estimativa teórica do modelo subjacente que gerou os dados, não como uma contagem literal exata das frequências isoladas.
