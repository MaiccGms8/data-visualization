Problema 10: Projeto integrador - Quatro visões com Dados "Gorjetas"

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url_tips = "https://raw.githubusercontent.com/Rogerio-mack/Visualizacao-de-Dados-em-Python/main/data/tips.csv"
tips = pd.read_csv(url_tips)

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
sns.histplot(tips['total_bill'], ax=axes[0,0], bins=20)
axes[0,0].set_title('Histograma')

sns.kdeplot(tips['total_bill'], fill=True, ax=axes[0,1])
axes[0,1].set_title('KDE')

sns.ecdfplot(tips['total_bill'], ax=axes[1,0])
axes[1,0].set_title('ECDF')

sns.boxplot(x=tips['total_bill'], ax=axes[1,1])
axes[1,1].set_title('Box-Plot')

plt.tight_layout()
plt.show()
```

a) a f) Análise do Total da Conta (total_bill): O centro situa-se perto de 18 dólares. A distribuição é assimétrica à direita com notória concentração na faixa de $10-$20, sem multimodalidade clara aparente. A dispersão estende-se severamente pelo eixo positivo e há existência comprovada de potenciais outliers acima de 40 dólares.

g) ECDF vs Histograma: A ECDF informa exatamente o percentil real atingido (ex: 80% das contas dão menos de $30), enquanto o histograma cru apenas fornece frequências absolutas por barra.

h) Box-plot vs KDE: O box-plot expõe os exatos limites IQR calculados, destacando a fronteira em que os pontos finais passam a ser quantificados como anómalos (outliers).

i) Acréscimo do Violin Plot: Retiraria o formato linear estático do box-plot, acrescentando os detalhes "gordos" nas densidades populacionais (quartis).

j) Apresentação para público leigo: O Histograma é o de compreensão imediata mais universal para massas devido ao cruzamento linear entre volume no "Y" vs montante no "X".

k) Conclusão: O cruzamento metodológico demonstra que o perfil do restaurante assenta numa média de valores baixos-medianos, caracterizando um serviço massificado. Eventos anómalos de grande conta, contudo, influenciam a assimetria global, requerendo atenção especial caso o analista se decida a utilizar apenas métricas médias simples de report.
