Problema 4: Atrasos de voos - box-plots e comparação

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/Rogerio-mack/Visualizacao-de-Dados-em-Python/main/data/flights_delays_2015.csv"
flights = pd.read_csv(url)

# a) e b) Filtragem das 5 companhias
top5_names = flights['name'].value_counts().head(5).index
flights_top5 = flights[flights['name'].isin(top5_names)].copy()
flights_top5['arr_delay'] = pd.to_numeric(flights_top5['arr_delay'], errors='coerce')

# c) e e) Box-plots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
sns.boxplot(data=flights_top5, x='name', y='arr_delay', ax=ax1)
ax1.set_title("Com fliers")
ax1.tick_params(axis='x', rotation=45)

sns.boxplot(data=flights_top5, x='name', y='arr_delay', showfliers=False, ax=ax2)
ax2.set_title("Sem fliers (showfliers=False)")
ax2.tick_params(axis='x', rotation=45)
plt.show()
```

d) Comparação visual: As medianas das cinco companhias estão muito próximas (perto de zero). No entanto, algumas apresentam bigodes maiores ou quantidades maciças de pontos muito afastados da caixa (outliers).

f) Ocultar vs Remover: Omitir fliers atua estritamente na renderização visual do gráfico no Seaborn/Matplotlib para melhorar o zoom da caixa. Os valores não são excluídos do DataFrame e as estatísticas mantêm-se iguais.

g) Comparação Interpretativa: Ao avaliar duas companhias (ex: Delta vs American Airlines), notamos que a mediana de ambas indica pontualidade geral, mas os dados sem fliers permitem ver que o IQR de uma é mais compacto. A existência de outliers maciços indica que, em situações de exceção (clima, avarias), as operações prolongam atrasos severos igualmente em ambas.
