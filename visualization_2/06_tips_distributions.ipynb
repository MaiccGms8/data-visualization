Problema 6: Gorjetas - box-plot, violin plot

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url_tips = "https://raw.githubusercontent.com/Rogerio-mack/Visualizacao-de-Dados-em-Python/main/data/tips.csv"
tips = pd.read_csv(url_tips)
tips["tip_pct"] = tips["tip"] / tips["total_bill"] * 100

# a, b e c) Box-plot e Violin Plot
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
sns.boxplot(data=tips, x='day', y='tip_pct', ax=ax1)
ax1.set_title("Box-plot")

sns.violinplot(data=tips, x='day', y='tip_pct', inner="quart", ax=ax2)
ax2.set_title("Violin plot (inner='quart')")
plt.show()
```

d) Informação predominante: O box-plot exibe quartis precisos e isola outliers matemáticos. O violin plot mostra o formato da distribuição e a densidade exata onde os dados se concentram.

e) e f) Concentração e Dispersão: As quintas ("Thur") mostram a distribuição mais comprimida em torno da mediana. Domingos e sábados ("Sun", "Sat") apresentam uma cauda alongada expressiva indicando dispersão acentuada nas porcentagens de gorjeta.

g) e h) Assimetria e Outliers: Todos os grupos apresentam assimetria à direita (limite no 0, mas podendo estender-se sem limite fixo). Sim, os outliers do box-plot refletem-se no violin plot na forma de topos longos e afilados muito distantes do centro, evidenciando as mesmas caudas longas.

i) Interpretação (Comparação dos dias): Percebe-se que durante o fim de semana o comportamento de consumo/gorjetas apresenta perfis mais mistos, com pessoas dispostas a dar percentagens muito altas, alargando a dispersão global. Por outro lado, o padrão dos dias úteis (como quinta-feira) indica comportamentos de consumo muito padronizados e regulares.
