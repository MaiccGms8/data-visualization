Problema 2: ECDF e probabilidades empíricas

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/Rogerio-mack/Visualizacao-de-Dados-em-Python/main/data/flights_delays_2015.csv"
arr_delay = pd.to_numeric(pd.read_csv(url)['arr_delay'], errors='coerce').dropna()

# a) e b) ECDF
x = np.sort(arr_delay)
y = np.arange(1, len(x) + 1) / len(x)

fig, ax = plt.subplots(figsize=(8, 5))
ax.step(x, y, label="ECDF Manual")
sns.ecdfplot(arr_delay, ax=ax, label="ECDF Seaborn", linestyle='--')
ax.legend()
plt.show()

# c) e e) Probabilidades e Quantis
print(f"P(X <= 0): {np.mean(arr_delay <= 0):.4f}")
print(f"P(X <= 15): {np.mean(arr_delay <= 15):.4f}")
print(f"P(X <= 30): {np.mean(arr_delay <= 30):.4f}")
print(f"P(X > 60): {np.mean(arr_delay > 60):.4f}")
print(f"Quantis (25%, 50%, 75%, 90%): {np.quantile(arr_delay, [0.25, 0.50, 0.75, 0.90])}")
```

d) Localização no gráfico: As probabilidades calculadas correspondem exatamente ao valor no eixo Y onde a curva interceta os valores de X (0, 15, 30 e 60).

f) Significado do quantil 90%: Significa que 90% dos voos registaram um atraso igual ou inferior ao valor obtido no cálculo. Os 10% restantes são os voos mais criticamente atrasados.

g) ECDF vs Histograma: O histograma acumulado depende da escolha arbitrária de bins, perdendo precisão exata. A ECDF é independente de classes, desenhando um degrau exato para cada observação isolada.
