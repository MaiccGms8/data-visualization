Problema 7: Pinguins Palmer - distribuições e agrupamentos

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url_penguins = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv"
penguins = pd.read_csv(url_penguins).dropna(subset=['body_mass_g', 'species', 'sex'])

# b a f) Representações Gráficas
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

sns.histplot(data=penguins, x='body_mass_g', ax=axes[0,0])
axes[0,0].set_title("Histograma Global")

sns.kdeplot(data=penguins, x='body_mass_g', hue='species', fill=True, ax=axes[0,1])
axes[0,1].set_title("KDE por Espécie")

sns.boxplot(data=penguins, x='species', y='body_mass_g', ax=axes[1,0])
axes[1,0].set_title("Box-plot por Espécie")

sns.violinplot(data=penguins, x='species', y='body_mass_g', hue='sex', split=True, ax=axes[1,1])
axes[1,1].set_title("Violin plot (Espécie + Sexo)")

plt.tight_layout()
plt.show()
```

g) Comparação das espécies: Gentoo apresenta um centro descolado à direita, indicando massas corporais muito superiores às demais, com dispersão própria. As espécies Adelie e Chinstrap apresentam formas, centros e dispersões parecidas, existindo grande sobreposição entre elas.

h) Massa corporal distribuída igualmente? Não, existem diferenças estatísticas gritantes entre a espécie Gentoo e o resto do grupo, sendo um claro fator de separação de classe.

i) Melhor representação: O Violin plot fragmentado (hue='sex' e split=True) revela-se o mais informativo, pois elucida não só a variação e densidade da espécie, mas responde perfeitamente à bimodalidade aparente causada pelo dimorfismo sexual da massa das aves.
