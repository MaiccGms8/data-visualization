Problema 1: Atrasos de voos (Histograma e escolha de classes)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/Rogerio-mack/Visualizacao-de-Dados-em-Python/main/data/flights_delays_2015.csv"
flights = pd.read_csv(url)

# a) Eliminação de valores ausentes
arr_delay = pd.to_numeric(flights['arr_delay'], errors='coerce').dropna()

# b) Estatísticas descritivas
print(f"N: {len(arr_delay)}")
print(f"Min: {arr_delay.min()} | Max: {arr_delay.max()}")
print(f"Média: {arr_delay.mean():.2f} | Mediana: {arr_delay.median()}")
print(f"Desvio-padrão: {arr_delay.std():.2f}")

# c) Construção dos histogramas
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for ax, b in zip(axes.flatten(), [20, 50, 100, 200]):
    sns.histplot(arr_delay, bins=b, ax=ax)
    ax.set_title(f'Histograma com bins={b}')
plt.tight_layout()
plt.show()
```

d) Comparação: O número de classes muda a percepção da distribuição. Poucos bins (20) simplificam em excesso e escondem padrões, enquanto muitos bins (200) introduzem ruído visual excessivo, fragmentando as contagens.

e) Assimetria: A distribuição é assimétrica à direita. Existe um limite rígido à esquerda (voos não chegam infinitamente adiantados), mas a cauda estende-se longamente para a direita (atrasos positivos).

f) Cauda longa: Sim. O significado prático é que, apesar da grande maioria dos voos se concentrar perto de zero, há uma pequena proporção de voos que sofre atrasos severos (várias horas).

g) Escolha: O histograma com bins=100 é o mais adequado, pois equilibra a suavização do ruído e a preservação de detalhes vitais da concentração ao redor do zero.
