**Objetivos:** 
- Construir e formatar um gráfico de linhas para séries temporais.
- Adicionar marcadores e anotações baseadas em condições matemáticas.
- Formatar o eixo de datas.
- Interpretar os dados temporalmente.

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Dados do problema
datas = pd.date_range("2026-01-01", periods=10, freq="MS")
atendimentos = [120, 128, 135, 132, 148, 160, 154, 169, 181, 195]

fig, ax = plt.subplots(figsize=(10, 5))

# a) e b) Gráfico de linhas utilizando marcadores circulares ('o')
ax.plot(datas, atendimentos, marker='o', linestyle='-', color='#1F77B4', linewidth=2, markersize=8)

# c) Destacar o mês de maior número de atendimentos utilizando annotate()
maior_atendimento = max(atendimentos)
indice_maior = atendimentos.index(maior_atendimento)
data_maior = datas[indice_maior]

ax.annotate(
    f"Pico: {maior_atendimento} atend.",
    xy=(data_maior, maior_atendimento),
    xytext=(datas[6], 190), # Desloca o texto para a esquerda (índice 6 = Julho)
    arrowprops=dict(facecolor='black', arrowstyle="->", lw=1.5),
    fontsize=11, fontweight='bold', color='#D62728'
)

# d) Incluindo título, rótulos e uma grade discreta
ax.set_title("Evolução Mensal do Número de Atendimentos (2026)", fontsize=14, pad=15)
ax.set_xlabel("Mês", fontsize=12)
ax.set_ylabel("Quantidade de Atendimentos", fontsize=12)
ax.grid(True, linestyle='--', alpha=0.5) # Grade discreta (alpha reduz a opacidade)

# e) Alterando o formato das datas do eixo X para uma apresentação mais legível
# O DateFormatter("%b/%Y") vai deixar no formato "Jan/2026", "Fev/2026"...
ax.xaxis.set_major_formatter(mdates.DateFormatter("%b/%Y"))

# Inclina as datas automaticamente para evitar sobreposição
fig.autofmt_xdate()

plt.show()
```

**f) Descreva a tendência observada:**

A série temporal apresenta uma clara tendência de crescimento (alta) ao longo dos 10 meses observados. Embora ocorram pequenas quedas pontuais (como de março para abril e de junho para julho), o padrão geral é de expansão contínua, saindo de 120 atendimentos em janeiro e alcançando um pico de 195 atendimentos no mês de outubro.

**g) Explique por que um gráfico de linhas é mais adequado que um gráfico de barras para destacar a evolução temporal nesse caso:**

O gráfico de linhas é superior para tempo porque a linha visualmente conecta os pontos, enfatizando a ideia de continuidade, fluxo e taxa de mudança (a inclinação da linha mostra se cresceu rápido ou devagar). 

O gráfico de barras, por outro lado, separa os dados em blocos isolados. Ele enfatiza a magnitude individual de cada mês como se fossem categorias independentes, quebrando a sensação fluida de "evolução" ao longo do tempo. Quando o eixo X possui uma ordem matemática contínua (como datas ou horas), a linha é quase sempre a melhor escolha.
