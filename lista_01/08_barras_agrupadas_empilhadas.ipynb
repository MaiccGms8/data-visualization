**Objetivos:** 
- Construir e diferenciar gráficos de barras agrupadas e empilhadas.
- Aplicar formatação completa (títulos, legendas e rótulos).
- Analisar os dados a partir das visualizações.
- Entender o caso de uso ideal para cada tipo de gráfico.

```python
import numpy as np
import matplotlib.pyplot as plt

# Dados do problema
equipes = ['A', 'B', 'C', 'D']
manha = np.array([35, 42, 31, 38])
tarde = np.array([28, 39, 44, 36])

# a) Construindo o gráfico de barras agrupadas
x = np.arange(len(equipes))
largura = 0.35 # Define a largura de cada barra

fig, ax = plt.subplots(figsize=(8, 5))

# Deslocamos as barras da manhã para a esquerda (- largura/2) e as da tarde para a direita (+ largura/2)
ax.bar(x - largura/2, manha, largura, label='Manhã', color='#4C78A8', edgecolor='black')
ax.bar(x + largura/2, tarde, largura, label='Tarde', color='#F58518', edgecolor='black')

# c) Formatação (Eixos, Título e Legenda)
ax.set_title("Solicitações Resolvidas por Equipe e Turno (Agrupado)", fontsize=14)
ax.set_xlabel("Equipe", fontsize=12)
ax.set_ylabel("Quantidade de Solicitações", fontsize=12)
ax.set_xticks(x)
ax.set_xticklabels(equipes)
ax.legend(title="Turno")

# Adicionando os valores em cima das barras
for container in ax.containers:
    ax.bar_label(container, padding=3)

ax.set_ylim(0, 55)
plt.show()
```

```python
# b) Construindo o gráfico de barras empilhadas
fig, ax = plt.subplots(figsize=(8, 5))

# A barra da manhã é desenhada normalmente. 
ax.bar(equipes, manha, label='Manhã', color='#4C78A8', edgecolor='black')
# A barra da tarde usa o parâmetro 'bottom' para começar exatamente onde a da manhã termina
ax.bar(equipes, tarde, bottom=manha, label='Tarde', color='#F58518', edgecolor='black')

# c) Formatação (Eixos, Título e Legenda)
ax.set_title("Solicitações Resolvidas por Equipe e Turno (Empilhado)", fontsize=14)
ax.set_xlabel("Equipe", fontsize=12)
ax.set_ylabel("Total de Solicitações", fontsize=12)
ax.legend(title="Turno")

ax.set_ylim(0, 95)
plt.show()
```

**d) Responda:**
- **Qual equipe resolveu mais solicitações no total?**  
  A Equipe B, com 81 solicitações (42 de manhã + 39 de tarde).
- **Em qual equipe a diferença entre os turnos foi maior?**  
  Na Equipe C, com uma diferença de 13 solicitações (44 à tarde contra 31 de manhã).
- **Qual turno apresentou maior total geral?**  
  O turno da Tarde, mas por uma margem mínima: 147 solicitações (contra 146 da Manhã).

---

**e) Para comparar diretamente manhã e tarde dentro de cada equipe, qual dos dois gráficos é melhor? Justifique.**

O gráfico de barras agrupadas é muito melhor para isso. Como ambas as barras (manhã e tarde) começam na mesma base (no eixo zero), nossos olhos conseguem comparar o comprimento das duas facilmente. No gráfico empilhado, a barra do turno da tarde flutua, começando em posições diferentes para cada equipe, o que dificulta muito a percepção visual de quem é maior.

**f) Para observar a composição do total de cada equipe, qual dos dois é mais adequado?**

O gráfico de barras empilhadas. Ele permite ver imediatamente a altura total (desempenho global da equipe) e, ao mesmo tempo, revela como esse total é "fatiado" (a proporção com a qual cada turno contribuiu para o resultado final). O gráfico agrupado não permite visualizar o tamanho do "Total" sem que tenhamos que fazer o cálculo mental somando as duas barras.
