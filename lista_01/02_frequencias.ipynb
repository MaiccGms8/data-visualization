**Objetivos:** 
- Calcular frequências absolutas, relativas e percentuais.
- Construir e comparar gráficos de colunas com valores absolutos e percentuais.
- Interpretar os resultados.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Dados fornecidos
sistemas = ["Windows", "Linux", "macOS", "ChromeOS", "Outro"]
frequencias = np.array([24, 12, 8, 4, 2])

# a) Calculando frequência relativa e o percentual
total_estudantes = frequencias.sum()
freq_relativa = frequencias / total_estudantes
percentuais = freq_relativa * 100

# b) Construindo a tabela completa com Pandas
tabela = pd.DataFrame({
    "Frequência Absoluta": frequencias,
    "Frequência Relativa": freq_relativa,
    "Percentual (%)": percentuais
}, index=sistemas) # Usando os sistemas como nome das linhas

# Exibindo a tabela formatada
display(tabela.round(2))
```

```python
# c) Gráfico de colunas das frequências absolutas
fig, ax = plt.subplots(figsize=(8, 5))
barras_abs = ax.bar(sistemas, frequencias, color='#4C78A8', edgecolor='black')

ax.set_title("Sistemas Operacionais Utilizados (Valores Absolutos)", fontsize=14)
ax.set_xlabel("Sistema Operacional", fontsize=12)
ax.set_ylabel("Quantidade de Estudantes (Frequência Absoluta)", fontsize=12)

# Adicionando os números exatos em cima das colunas
for barra in barras_abs:
    altura = barra.get_height()
    ax.text(barra.get_x() + barra.get_width()/2, altura + 0.5, f'{int(altura)}', ha='center')

ax.set_ylim(0, 30)
plt.show()
```

```python
# d) Gráfico de colunas com os percentuais
fig, ax = plt.subplots(figsize=(8, 5))
barras_perc = ax.bar(sistemas, percentuais, color='#F58518', edgecolor='black')

ax.set_title("Sistemas Operacionais Utilizados (Distribuição Percentual)", fontsize=14)
ax.set_xlabel("Sistema Operacional", fontsize=12)
ax.set_ylabel("Percentual da Turma (%)", fontsize=12)

# Adicionando as porcentagens em cima das colunas
for barra in barras_perc:
    altura = barra.get_height()
    ax.text(barra.get_x() + barra.get_width()/2, altura + 1, f'{altura:.0f}%', ha='center')

ax.set_ylim(0, 60)
plt.show()
```

**e) Qual das duas versões você considera mais adequada para comunicar a distribuição da turma? Justifique.**

A versão em **percentuais (letra d)** é muito mais adequada. Quando queremos comunicar a "distribuição" de algo, o leitor quer entender proporções (as "fatias do bolo"). Ao usar valores absolutos (ex: 24 alunos), o leitor é obrigado a calcular mentalmente a relação desse número com o total (que nem sempre ele sabe de cabeça). O percentual (ex: 48%) já entrega essa informação processada, mostrando imediatamente que quase metade da turma usa Windows, independentemente de a turma ter 50 ou 500 alunos.
