**Objetivos:** 
- Demonstrar como a manipulação do eixo vertical afeta a percepção visual.
- Entender as regras de uso da linha de base zero (baseline).
- Definir boas práticas para construção de gráficos éticos e precisos.

```python
import matplotlib.pyplot as plt

empresas = ["A", "B", "C", "D"]
lucros = [94, 98, 101, 105]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# a) Gráfico 1: Eixo vertical iniciando em 0 (
axes[0].bar(empresas, lucros, color='#4C78A8', edgecolor='black')
axes[0].set_title("Lucro Anual (Eixo iniciando em 0)", fontsize=14)
axes[0].set_ylabel("Lucro (Milhares de R$)", fontsize=12)
axes[0].set_ylim(0, 120) # Forçando o eixo a começar do zero

# b) Gráfico 2: Eixo vertical iniciando em 90 
axes[1].bar(empresas, lucros, color='#D62728', edgecolor='black')
axes[1].set_title("Lucro Anual (Eixo truncado em 90)", fontsize=14)
axes[1].set_ylabel("Lucro (Milhares de R$)", fontsize=12)
axes[1].set_ylim(90, 110) # Truncando o eixo no 90

for i in range(2):
    for barra in axes[i].patches:
        altura = barra.get_height()
        offset = 1 if i == 0 else -2
        color = 'black' if i == 0 else 'white'
        axes[i].text(barra.get_x() + barra.get_width()/2, altura + offset, 
                     f'{int(altura)}', ha='center', color=color, fontweight='bold')

plt.tight_layout()
plt.show()
```

**c) Compare visualmente as duas figuras.**

No primeiro gráfico, as quatro empresas parecem ter um desempenho muito semelhante, com barras de alturas quase iguais. No segundo gráfico, a barra da empresa D parece ser gigantesca, visualmente quase quatro vezes maior que a barra da empresa A, sugerindo uma diferença brutal de lucros.

---

**d) Explique por que a segunda versão pode transmitir uma impressão exagerada das diferenças entre as empresas.**

Ao cortar o eixo em 90, a parte desenhada da barra A vale 4 (94-90) e a da D vale 15 (105-90). Visualmente, 15 é quase 4 vezes maior que 4. Truncar a base destrói a representação proporcional da barra, gerando uma manipulação enganosa.

**e) Em que situações alterar os limites dos eixos pode ser justificável?**

É justificável truncar eixos principalmente em Gráficos de Linha e Gráficos de Dispersão.

**f) Escreva duas recomendações de boas práticas relacionadas ao uso de escalas em gráficos.**

1. Gráficos de barras horizontais ou verticais devem começar a escala no zero. 
2. Se você precisar usar um eixo truncado (em linhas/dispersão) ou escalas não lineares (como a logarítmica), deixe isso extremamente claro para o leitor. Use marcas de interrupção no eixo ou coloque um aviso explícito no título ou subtítulo, garantindo que o leitor saiba que o eixo não começa do zero.
