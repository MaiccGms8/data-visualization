**Objetivos:** 
- Produzir uma visualização propositalmente ruim, ferindo os princípios de design de dados.
- Construir a versão corrigida (Storytelling com Dados).
- Explicar as correções aplicadas.

**Conjunto de Dados Escolhido:** Percentual de vagas de Ciência de Dados que exigem determinadas linguagens de programação.

```python
import matplotlib.pyplot as plt

# Dados do problema (Propositalmente fora de ordem)
linguagens = ['Java', 'Python', 'Scala', 'R', 'SQL']
porcentagens = [25, 85, 15, 40, 75]

# 1. A VERSÃO PROPOSITALMENTE RUIM
fig, ax = plt.subplots(figsize=(7, 5))

# Excesso de cores sem significado (arco-íris neon)
cores_ruins = ['#FF00FF', '#00FFFF', '#00FF00', '#FFFF00', '#FF0000']

# Plotando as barras
ax.bar(linguagens, porcentagens, color=cores_ruins, label=linguagens)

# Título pouco informativo
ax.set_title("Gráfico 1 - Linguagens")

# Escala inadequada (truncando o eixo Y e exagerando as diferenças)
# Sem rótulo informando se é quantidade, porcentagem ou dinheiro
ax.set_ylim(10, 90)

# Legenda totalmente desnecessária (o eixo X já tem os nomes)
ax.legend(loc='upper right')

# Grade excessiva e distrativa
ax.grid(color='black', linestyle='-', linewidth=1)

plt.show()
```

```python
import numpy as np

# Reorganizando os dados (Ordem decrescente)
dados = {'Python': 85, 'SQL': 75, 'R': 40, 'Java': 25, 'Scala': 15}
ling_ordenadas = list(dados.keys())
perc_ordenadas = list(dados.values())

# 2. A VERSÃO CORRIGIDA (BOAS PRÁTICAS)
fig, ax = plt.subplots(figsize=(8, 5))

# Cor estratégica: Destaque para as duas principais, cinza neutro para o resto
cores_boas = ['#1F77B4', '#1F77B4', '#CCCCCC', '#CCCCCC', '#CCCCCC']

# Barras horizontais para facilitar a leitura das categorias
barras = ax.barh(ling_ordenadas, perc_ordenadas, color=cores_boas, edgecolor='none')

# Invertendo o eixo Y para a maior barra ficar no topo
ax.invert_yaxis()

# Título informativo com a conclusão da análise
ax.set_title("Python e SQL dominam as exigências em vagas de Dados", 
             fontsize=14, fontweight='bold', loc='left', pad=15)
ax.set_xlabel("Percentual de vagas que exigem a linguagem (%)", fontsize=11, color='#333333')

# Escala honesta começando do ZERO e indo até 100%
ax.set_xlim(0, 100)

# Removendo bordas desnecessárias (Decluttering)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_color('#CCCCCC')

# Removendo os ticks do eixo Y (os tracinhos) para deixar mais limpo
ax.tick_params(left=False)

# Adicionando rótulos de dados diretos nas barras
for barra in barras:
    ax.text(barra.get_width() + 2, barra.get_y() + barra.get_height()/2,
            f'{barra.get_width()}%', va='center', fontsize=11, fontweight='bold', color='#333333')

plt.show()
```

### Quais problemas foram corrigidos na segunda versão?

1. **Escala enganosa:** Na versão ruim, o eixo Y começava em 10 e terminava em 90. Isso fazia a linguagem Scala (15%) parecer praticamente invisível e Python parecer infinitamente maior. Na versão corrigida, o eixo começa rigorosamente no 0 e vai até 100, respeitando a proporção matemática real (Princípio da integridade visual).
2. **Excesso de cores e Legenda desnecessária:** A versão ruim parecia um arco-íris, onde as cores não significavam nada e a legenda repetia os nomes que já estavam no eixo X. Na versão corrigida, a legenda foi removida e as cores foram usadas para destacar (azul para as linguagens líderes, cinza neutro para o restante).
3. **Categorias fora de ordem:** A versão ruim exibia as linguagens de forma aleatória, forçando o leitor a ficar procurando quem era o primeiro ou o último. A versão corrigida ordenou os dados **do maior para o menor**.
4. **Barras Horizontais:** Mudamos de colunas verticais para barras horizontais. Em gráficos com nomes de categorias textuais, as barras horizontais evitam que o leitor tenha que entortar o pescoço para ler textos sobrepostos ou inclinados.
5. **Título e Unidades:** O título "Gráfico 1" foi substituído por uma manchete informativa que já entrega a resposta ao leitor ("Python e SQL dominam..."). Além disso, o eixo X agora explica claramente que os números se referem ao "Percentual de vagas (%)".
6. **Remoção de ruído (Decluttering):** A grade preta e pesada foi totalmente removida, pois os rótulos de dados diretos (ex: 85%) já eliminam a necessidade de linhas de grade para consultar os valores.
