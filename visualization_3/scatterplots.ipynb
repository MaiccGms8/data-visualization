Scatterplots com Penguins

```python
import seaborn as sns
import matplotlib.pyplot as plt

penguins = sns.load_dataset("penguins")

# Passos 1, 2 e 3
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# V1: Básico
sns.scatterplot(data=penguins, x="flipper_length_mm", y="body_mass_g", ax=axes[0])
axes[0].set_title("V1: Básico")

# V2: + Espécie
sns.scatterplot(data=penguins, x="flipper_length_mm", y="body_mass_g", hue="species", ax=axes[1])
axes[1].set_title("V2: hue='species'")

# V3: + Espécie e Sexo
sns.scatterplot(data=penguins, x="flipper_length_mm", y="body_mass_g", hue="species", style="sex", ax=axes[2])
axes[2].set_title("V3: hue='species' e style='sex'")

plt.tight_layout()
plt.show()
```

Comparação (4): 

O gráfico V1 mostra apenas uma forte tendência positiva geral. O V2 introduz cores e revela três clusters perfeitamente separados por espécie (mostrando que a espécie Gentoo tem peso e nadadeiras nitidamente maiores). O V3 usa formas para identificar os sexos, indicando que machos tendem a ser maiores que fêmeas em cada espécie, porém insere muitos elementos simultâneos.

Escolha (5): 

A versão V2 (apenas com hue="species") é a mais informativa sem ficar sobrecarregada. O uso simultâneo de cores e formas na V3 gera poluição visual, exigindo alto esforço cognitivo para separar os símbolos de sexos no meio da sobreposição dos pontos.
