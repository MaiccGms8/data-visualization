Pairplot e Correlação

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Carregar o dataset antes de usá-lo
penguins = sns.load_dataset("penguins")

# Filtrar as colunas numéricas
cols = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]
peng_num = penguins[cols].dropna()

# 1. Pairplot
sns.pairplot(peng_num)
plt.show()

# 2. Matriz de correlação
corr_matrix = peng_num.corr()

# 3. Heatmap
plt.figure(figsize=(6, 5))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Heatmap - Penguins")
plt.show()

# 5. Verificação visual da maior correlação
sns.scatterplot(data=penguins, x="flipper_length_mm", y="body_mass_g")
plt.title("Verificação: Flipper Length vs Body Mass")
plt.show()
```

Maior correlação absoluta (4): 

Ocorre entre flipper_length_mm e body_mass_g, alcançando aproximadamente 0.87 de correlação positiva.

Padrão Visual (5): 

O gráfico de dispersão dessas duas variáveis confirma a matemática: os pontos acompanham perfeitamente um eixo diagonal ascendente muito estreito, evidenciando a forte proporcionalidade direta (alta massa corporal reflete em grandes nadadeiras).
