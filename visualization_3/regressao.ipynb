Regressão com Tips

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

# 1. regplot
plt.figure()
sns.regplot(data=tips, x="total_bill", y="tip")
plt.title("Regplot: Total Bill vs Tip")
plt.show()

# 2. lmplot com hue
sns.lmplot(data=tips, x="total_bill", y="tip", hue="time")
plt.title("Lmplot: hue='time'")
plt.show()

# 4. Facetas por day
sns.lmplot(data=tips, x="total_bill", y="tip", hue="time", col="day", col_wrap=2)
plt.show()
```

Comparação das linhas (3): 

O regplot único mostra uma única reta positiva de ajuste. Ao rodar o lmplot com hue="time", vemos que as inclinações das linhas para Almoço (Lunch) e Jantar (Dinner) seguem ângulos bem parecidos, indicando que a taxa média calculada da gorjeta acompanha a conta de forma semelhante nos dois períodos.

Interpretação s/ causalidade (5): 

Nota-se uma associação positiva robusta entre as variáveis — refeições que registram gastos maiores tendem a vir acompanhadas de gorjetas de maior valor numérico. Esse padrão se repete em quase todos os dias analisados. Contudo, não é possível afirmar com isso que uma conta cara seja a causa exclusiva do aumento da gorjeta (as variáveis podem estar sendo movidas conjuntamente por fatores como número de pessoas sentadas à mesa, duração da estadia ou qualidade do prato).
