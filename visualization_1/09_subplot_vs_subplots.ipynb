**Objetivos:** 
- Criar grades de gráficos (painéis) com múltiplas visualizações.
- Entender a diferença prática entre `plt.subplot()` e `plt.subplots()`.
- Ajustar o espaçamento da figura e formatar cada `Axes` individualmente.

```python
import numpy as np
import matplotlib.pyplot as plt

# Dados do problema
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = x
y4 = x**2
```

```python
# a e b) Utilizando plt.subplot() para construir uma grade 2x2
plt.figure(figsize=(10, 7))

# Gráfico 1 (Seno)
plt.subplot(2, 2, 1) # 2 linhas, 2 colunas, gráfico 1 (canto superior esquerdo)
plt.plot(x, y1, color='#1F77B4')
plt.title("Seno")

# Gráfico 2 (Cosseno)
plt.subplot(2, 2, 2) # gráfico 2 (canto superior direito)
plt.plot(x, y2, color='#FF7F0E')
plt.title("Cosseno")

# Gráfico 3 (Função Linear)
plt.subplot(2, 2, 3) # gráfico 3 (canto inferior esquerdo)
plt.plot(x, y3, color='#2CA02C')
plt.title("Função Linear")

# Gráfico 4 (Função Quadrática)
plt.subplot(2, 2, 4) # gráfico 4 (canto inferior direito)
plt.plot(x, y4, color='#D62728')
plt.title("Função Quadrática")

plt.tight_layout() # Ajusta os espaçamentos para evitar sobreposição de textos
plt.show()
```

```python
# c, d, e, f) Repetindo o exercício com fig, axes = plt.subplots(2, 2)
# O parâmetro constrained_layout=True já faz o trabalho do tight_layout() automaticamente
fig, axes = plt.subplots(2, 2, figsize=(10, 7), constrained_layout=True)

# Acessando cada Axes como se fosse uma matriz [linha, coluna] (o índice começa no 0)
# axes[0, 0] é a 1ª linha, 1ª coluna
axes[0, 0].plot(x, y1, color='#1F77B4')
axes[0, 0].set_title("Seno")

axes[0, 1].plot(x, y2, color='#FF7F0E')
axes[0, 1].set_title("Cosseno")

axes[1, 0].plot(x, y3, color='#2CA02C')
axes[1, 0].set_title("Função Linear")

axes[1, 1].plot(x, y4, color='#D62728')
axes[1, 1].set_title("Função Quadrática")

plt.show()
```

**g) Explique por que subplots() tende a ser mais conveniente quando a figura possui muitos gráficos:**

A função `plt.subplots()` é muito mais limpa e programável. Com o `plt.subplot()`, você precisa ativar a região do gráfico manualmente repetidas vezes (`plt.subplot(3, 3, 1)`, depois o 2, etc.), o que deixa o código longo e difícil de gerenciar. Com o `plt.subplots()`, a biblioteca devolve uma matriz Numpy (o objeto `axes`). Isso permite que você acesse cada gráfico pelo seu índice (ex: `axes[1, 2]`) e, mais importante, permite usar um loop `for` para preencher dezenas de gráficos em poucas linhas de código.

**h) Explique a função de sharex=True e sharey=True:**

Esses parâmetros forçam todos os gráficos do painel a **compartilharem a mesma escala (limites) no eixo X e/ou no eixo Y**. 
Por padrão, o Matplotlib autoajusta os limites de cada `Axes` independentemente com base nos dados que ele contém. Quando você quer comparar a magnitude de duas séries diferentes em gráficos lado a lado, usar `sharey=True` impede que a escala de um fique enganosa em relação ao outro, garantindo que 1 cm de altura em um gráfico signifique o exato mesmo valor no gráfico vizinho.
