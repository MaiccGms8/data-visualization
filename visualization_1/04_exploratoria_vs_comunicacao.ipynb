**Objetivos:** 
- Criar um gráfico rápido para exploração pessoal.
- Identificar padrões nos dados.
- Transformar o gráfico exploratório em uma visualização polida para comunicação, destacando a mensagem principal.

```python
import matplotlib.pyplot as plt

meses = ["Jan", "Fev", "Mar", "Abr", "Mai", "Jun", "Jul", "Ago", "Set", "Out", "Nov", "Dez"]
acessos = [1200, 1350, 1280, 1420, 1510, 1650, 1580, 1720, 1900, 2100, 2350, 2600]

# a) Visualização exploratória simples
plt.plot(meses, acessos)
plt.show()
```

**b) Identifique pelo menos três aspectos que podem ser percebidos no gráfico:**

1. Tendência geral de alta: 
2. Quedas pontuais.
3. Aceleração no último trimestre.

```python
# c) e d) Gráfico destinado à comunicação

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(meses, acessos, color='#4C78A8', linewidth=2.5, marker='o')

ax.plot(meses[-1], acessos[-1], marker='o', color='#D62728', markersize=10)

ax.set_title("Acessos à plataforma mais que dobraram em 2026", fontsize=16, loc='left', pad=15, fontweight='bold')
ax.set_ylabel("Quantidade de Acessos mensais", fontsize=11, color='#333333')

ax.annotate(
    "Pico",
    xy=(meses[-1], acessos[-1]), 
    xytext=("Ago", 2400),        
    arrowprops=dict(arrowstyle="->", color='#D62728', lw=1.5),
    fontsize=11, color='#D62728', fontweight='bold'
)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

ax.spines['left'].set_color('#CCCCCC')
ax.spines['bottom'].set_color('#CCCCCC')

ax.tick_params(bottom=False, left=False)


ax.grid(axis='y', linestyle='--', alpha=0.4)

plt.show()
```
