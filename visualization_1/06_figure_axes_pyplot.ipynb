**Objetivos:** 
- Compreender a diferença entre a interface funcional (`pyplot`) e a orientada a objetos (`Figure` e `Axes`).
- Adicionar formatações básicas a um gráfico (marcadores, linhas, títulos).
- Entender o dimensionamento de figuras (resolução e tamanho).

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [3, 5, 4, 8, 10]

# a) Construindo o gráfico utilizando a interface pyplot
plt.plot(x, y)
plt.show()
```

```python
# b) Construindo o mesmo gráfico com a interface orientada a objetos
fig, ax = plt.subplots(figsize=(8, 5))

# c) Acrescentando as formatações pedidas
ax.plot(x, y, marker='o', linestyle='--', color='#2CA02C', linewidth=2, markersize=8)

# Acrescentando títulos e rótulos (Letra C continuada)
ax.set_title("Evolução da Variável Y em relação a X", fontsize=14)
ax.set_xlabel("Eixo X (Variável Independente)", fontsize=12)
ax.set_ylabel("Eixo Y (Variável Dependente)", fontsize=12)

plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

**d) Explique a diferença entre Figure, Axes e pyplot:**

- **Figure:** É o "container" geral ou a "moldura" do quadro. É a janela principal que abriga tudo. Uma `Figure` pode conter um gráfico, nenhum gráfico ou até mesmo dezenas de gráficos menores dentro dela.
- **Axes:** É o gráfico em si. É a região real de plotagem (o retângulo que contém os eixos X e Y, os dados desenhados, os rótulos e as marcações). Uma `Figure` possui um ou mais `Axes`. *(Atenção: Axes não é o plural de Axis/Eixo, é o nome do objeto da área de plotagem).*
- **pyplot:** É um módulo do Matplotlib que funciona como uma "máquina de estados". Ele gerencia automaticamente a `Figure` e o `Axes` atuais sem que você precise nomeá-los. É ótimo para gráficos rápidos, mas menos flexível para painéis complexos do que a abordagem orientada a objetos.

```python
# e) Executando o comando
fig = plt.figure()
print(fig)
```

**Explicação da saída do comando anterior:**
A saída será algo como `<Figure size 640x480 with 0 Axes>`. Isso significa que o Matplotlib criou o "container" (a moldura da figura) com o tamanho padrão de 640x480 pixels na memória, porém ela está completamente vazia. O aviso `0 Axes` indica que nenhuma região de plotagem (nenhum gráfico) foi adicionada dentro dela ainda.

**f) Explique a relação entre figsize, dpi e tamanho final da imagem em pixels:**

O tamanho final de uma imagem gerada pelo Matplotlib é um cálculo matemático simples envolvendo esses três elementos:

1. **figsize:** Define as dimensões físicas desejadas para a figura em polegadas (inches), no formato `(largura, altura)`. Exemplo: `figsize=(8, 5)`.
2. **dpi (Dots Per Inch):** Significa "Pontos por Polegada" e define a densidade de pixels (resolução). O padrão do Matplotlib geralmente é `100 dpi`.
3. **Tamanho final em pixels:** É o resultado da multiplicação matemática do `figsize` pelo `dpi`.

**Fórmula:**
- `Pixels de Largura = largura (figsize) × dpi`
- `Pixels de Altura = altura (figsize) × dpi`

*Exemplo:* Uma figura com `figsize=(8, 5)` e `dpi=100` terá um tamanho final de `800x500 pixels` (8 * 100 e 5 * 100).
